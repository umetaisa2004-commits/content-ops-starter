```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

# 画像読み込み
path = "/content/image0012.tif"
img = cv2.imread(path)

if img is None:
    print("画像を読み込めませんでした。パスを確認してください。")
else:
    # Step 1: グレースケール化
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Step 2: 2値化
    ret, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

    # Step 3: ノイズ除去（オープニング処理）
    kernel = np.ones((3,3), np.uint8)
    opening = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel, iterations=2)

    # 【前準備】白黒反転と画像サイズの縮小表示
    opening_inv = cv2.bitwise_not(opening)
    h, w = opening_inv.shape
    resized_img = cv2.resize(opening_inv, (w // 2, h // 2))
    print("【前準備】フロックを白に反転")
    cv2_imshow(resized_img)

    # Step 4: 輪郭抽出
    contours, hierarchy = cv2.findContours(opening_inv, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    print(f"抽出された粒子の数: {len(contours)} 個")

    # Step 5: 粒径（等価円直径）の計算と分布表示
    diameters = []
    for cnt in contours:
        area = cv2.contourArea(cnt)
        if area > 10:  # 面積10ピクセル以下のノイズを除外
            diameter = 2 * np.sqrt(area / np.pi)
            diameters.append(diameter)

    # ヒストグラムの表示
    plt.figure(figsize=(8, 5))
    plt.hist(diameters, bins=40, color='teal', alpha=0.75, edgecolor='black')
    plt.title("Floc Size Distribution in Taylor-Couette Flow")
    plt.xlabel("Equivalent Diameter (pixels)")
    plt.ylabel("Frequency")
    plt.grid(axis='y', alpha=0.5)
    plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# --- 1. 前準備（画像の読み込みと二値化） ---
path = "/content/image0010.tif"
img = cv2.imread(path)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 【追加】微細なモヤを滑らかにして、フロック同士の不自然な結合を防ぐ
blurred = cv2.GaussianBlur(gray, (5, 5), 0)

# 【修正】画像の特徴に合わせて二値化（フロックが白、背景が黒になるよう自動判別）
ret, thresh = cv2.threshold(blurred, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

# フロックが黒くなってしまう画像の場合は、上の行を cv2.THRESH_BINARY_INV に変更してください

# 【修正】白いフロックに対してオープニング処理をかけ、細い繋がりを切断する
kernel = np.ones((3,3), np.uint8)
# iterationsを3に増やし、重なりの切り離し強度を上げています
flocs_mask = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel, iterations=3)

# --- 2. 輪郭抽出と粒径計算（μm変換） ---
contours, hierarchy = cv2.findContours(flocs_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
diameters_um = []

for cnt in contours:
    area = cv2.contourArea(cnt)
    if area > 50:
        diameter_px = 2 * np.sqrt(area / np.pi)
        diameter_um = diameter_px / 0.81
        diameters_um.append(diameter_um)

d_array = np.array(diameters_um)
total_count = len(d_array)

# --- 3. グラフの横並び描画 ---
if total_count > 0:
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))

    axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    axes[0].set_title("Original Image")
    axes[0].axis('off')

    # 修正したマスク画像を表示
    axes[1].imshow(flocs_mask, cmap='gray')
    axes[1].set_title("Binarized (Flocs = White)")
    axes[1].axis('off')

    weights = np.ones(total_count) / total_count * 100
    axes[2].hist(d_array, bins=40, weights=weights, color='teal', alpha=0.75, edgecolor='black')
    axes[2].set_title("Floc Size Distribution (%)")
    axes[2].set_xlabel("Equivalent Diameter (µm)")
    axes[2].set_ylabel("Percentage (%)")
    axes[2].grid(axis='y', alpha=0.5)

    c_under100 = np.sum(d_array < 100)
    c_100_200  = np.sum((d_array >= 100) & (d_array < 200))
    c_200_500  = np.sum((d_array >= 200) & (d_array < 500))
    c_over500  = np.sum(d_array >= 500)

    stats_text = (
        f"[ Total Flocs: {total_count} ]\n"
        f"< 100 µm   : {c_under100:3} ({c_under100/total_count*100:5.1f}%)\n"
        f"100-200 µm : {c_100_200:3} ({c_100_200/total_count*100:5.1f}%)\n"
        f"200-500 µm : {c_200_500:3} ({c_200_500/total_count*100:5.1f}%)\n"
        f">= 500 µm  : {c_over500:3} ({c_over500/total_count*100:5.1f}%)"
    )

    # 【修正】表の配置を右上に変更 (x=0.95, y=0.95, 右・上揃え)
    axes[2].text(0.95, 0.95, stats_text, transform=axes[2].transAxes,
                 ha="right", va="top", fontsize=10,
                 bbox=dict(boxstyle="round,pad=0.5", facecolor="white", edgecolor="gray", alpha=0.9))

    plt.tight_layout()
    plt.show()
else:
    print("条件を満たすフロックが見つかりませんでした。")
```
